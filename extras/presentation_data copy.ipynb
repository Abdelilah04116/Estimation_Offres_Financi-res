- **La description des colonnes de la table de données**

- **date_heure_limite_remise_plis**  
  🔹 Date et heure exactes où les soumissions doivent être déposées.  
  🔹 Sert à analyser les délais de réponse et la réactivité des entreprises.  

- **reference**  
  🔹 Code ou identifiant unique attribué à l’appel d’offre par l’organisme émetteur.  
  🔹 Permet un suivi administratif et facilite les recherches.  

- **objet**  
  🔹 Description courte du but ou de la nature de l’appel d’offre.  
  🔹 Sert à comprendre rapidement la finalité du projet.  

- **acheteur_public**  
  🔹 Administration ou entité publique responsable de l’appel d’offre.  
  🔹 Utile pour identifier l’émetteur et établir un historique.  

- **type_annonce**  
  🔹 Nature de la publication (appel d’offre, attribution, annulation…).  
  🔹 Permet de filtrer selon l’état du processus.  

- **procedure**  
  🔹 Méthode de passation (ouverte, restreinte, négociée…).  
  🔹 Indique le cadre juridique et administratif de la consultation.  

- **categorie_principale**  
  🔹 Domaine principal : Travaux, Fournitures, Services, etc.  
  🔹 Sert à segmenter les projets par nature.  

- **allotissement**  
  🔹 Indique si le marché est divisé en lots.  
  🔹 Permet de savoir si des entreprises peuvent soumissionner sur une partie du projet.  

- **lieu_execution**  
  🔹 Localisation géographique prévue pour la réalisation du projet.  
  🔹 Utile pour les analyses spatiales.  

- **domaines_activite**  
  🔹 Secteurs économiques ou techniques concernés.  
  🔹 Permet de regrouper les appels d’offres par filière.  

- **agrements**  
  🔹 Certificats ou qualifications obligatoires pour participer.  
  🔹 Sert à analyser les barrières à l’entrée.  

- **documents_annexes**  
  🔹 Documents annexes fournis par l’acheteur (prospectus, notices…).  
  🔹 Contiennent souvent des informations techniques détaillées.  

- **reunion**  
  🔹 Informations sur une éventuelle réunion d’information ou de clarification.  
  🔹 Utile pour comprendre le niveau d’accompagnement fourni.  

- **visites_lieux**  
  🔹 Informations sur les visites préalables des sites concernés.  
  🔹 Sert à identifier les contraintes logistiques.  

- **variante**  
  🔹 Indique si des propositions alternatives aux spécifications sont autorisées.  
  🔹 Permet de voir la flexibilité de l’acheteur.  

- **contact_administratif**  
  🔹 Coordonnées de la personne ou du service chargé du suivi administratif.  
  🔹 Utile pour établir un réseau ou poser des questions.  

- **ref_consultation**  
  🔹 Identifiant interne du dossier de consultation.  
  🔹 Sert pour la traçabilité dans les systèmes internes.  

- **org_acronyme**  
  🔹 Acronyme officiel de l’organisation émettrice.  
  🔹 Permet un traitement plus rapide dans les analyses.  

- **financial_data_json**  
  🔹 Données financières structurées au format JSON (budget, montant estimatif, etc.).  
  🔹 Crucial pour les analyses budgétaires.  

- **error**  
  🔹 Champ contenant les éventuelles erreurs ou anomalies détectées lors de la collecte.  
  🔹 Permet de filtrer et nettoyer les données.
